```python
import pandas as pd 
import numpy as np 
df = pd.DataFrame({ 
'A': [10, np.nan, 30], 
'B': [np.nan, 50, 60] 
}) 
print("Original DataFrame with NaN:") 
print(df) 
df_filled = df.fillna(0) 
print("DataFrame after Replacing NaN with 0:") 
print(df_filled)
```
