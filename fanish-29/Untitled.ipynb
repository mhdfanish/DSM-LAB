```python
import numpy as np

x=np.array([[2,3],[4,5]])
U, S, VT = np.linalg.svd(x)
n_comp=2
x_recon=np.dot(U[:,:n_comp],np.dot(np.diag(S[:n_comp]),VT[:n_comp,:]))
print("Oringinal matrix:")
print(x)
print("Reconstructed matrix:")
print(x_recon)
```
